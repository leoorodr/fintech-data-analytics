```sql
SELECT COUNT(*) AS total_clientes
FROM fintech.bronze.clientes;

```

```sql
SELECT COUNT(*) AS total_transacoes
FROM fintech.bronze.transacoes;
```

```sql
SELECT SUM(valor) AS total_valor_movimentado
FROM fintech.bronze.transacoes
```

```sql
SELECT avg(valor)AS ticket_medio
FROM fintech.bronze.transacoes
```

```sql
SELECT estado, COUNT(*) AS quantidade_clientes
FROM fintech.bronze.clientes
GROUP BY estado
ORDER BY quantidade_clientes DESC;
```

```sql
SELECT p.nome_produto,
    COUNT(*) AS quatidade_contratos
    FROM fintech.bronze.contratos c
    INNER JOIN fintech.bronze.produtos p
        ON c.id_produto = p.id_produto
    GROUP BY p.nome_produto 
    ORDER BY quatidade_contratos DESC;
```

```sql
SELECT
    faixa_renda,
    AVG(score_credito) AS score_medio
FROM fintech.bronze.clientes
GROUP BY faixa_renda;
```

```sql
SELECT
    tipo,
    COUNT(*) AS quantidade_transacoes,
    SUM(valor) AS valor_movimentado
FROM fintech.bronze.transacoes
GROUP BY tipo
ORDER BY valor_movimentado DESC;
```

```sql
SELECT
    id_cliente,
    SUM(valor) AS total_movimentado
FROM fintech.bronze.transacoes
GROUP BY id_cliente
ORDER BY total_movimentado DESC
LIMIT 10;
```

```sql
--Quantidade de transações que cada cliente faz em cada tipo de transação

SELECT 
    id_cliente,
    tipo,
    COUNT(*) AS quantidade_transacoes
FROM fintech.gold.fato_transacoes
GROUP BY id_cliente, tipo
ORDER BY quantidade_transacoes DESC
LIMIT 15;
```
