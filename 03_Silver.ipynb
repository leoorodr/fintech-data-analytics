```sql
DESCRIBE fintech.bronze.clientes;
```

```sql
DESCRIBE fintech.bronze.transacoes;
```

```sql
CREATE OR REPLACE TABLE fintech.silver.clientes AS
SELECT
    CAST(id_cliente AS INT) AS id_cliente,
    nome,
    sexo,
    idade,
    cidade,
    estado,
    CAST(data_cadastro AS DATE) AS data_cadastro,
    faixa_renda,
    CAST(score_credito AS INT) AS score_credito
FROM fintech.bronze.clientes
```

```sql
SELECT *
FROM fintech.silver.clientes
LIMIT 10;
```

```sql
CREATE OR REPLACE TABLE fintech.silver.produtos AS
SELECT
    CAST(id_produto AS INT) AS id_produto,
    nome_produto,
    categoria
FROM fintech.bronze.produtos
```

```sql
CREATE OR REPLACE TABLE fintech.silver.transacoes AS
SELECT 
    CAST(id_transacao AS INT) AS id_transacao,
    CAST(id_cliente AS INT) AS id_cliente,
    CAST(data AS DATE) AS data,
    tipo,
    CAST(valor AS DECIMAL(18,2)) AS valor
FROM fintech.bronze.transacoes
```

```sql
CREATE OR REPLACE TABLE fintech.silver.atendimentos AS
SELECT
    CAST(id_atendimento AS INT) AS id_atendimento,
    CAST(id_cliente AS INT) AS id_cliente,
    CAST(data AS DATE) AS data,
    canal,
    motivo,
    resolvido
FROM fintech.bronze.atendimentos
```

```sql
CREATE OR REPLACE TABLE fintech.silver.cancelamentos AS

SELECT 
    CAST(id_cancelamento AS INT) AS id_cancelamento,
    CAST(id_cliente AS INT) AS id_cliente,
    CAST(data_cancelamento AS DATE) AS data_cancelamento,
    motivo
FROM fintech.bronze.cancelamentos
```

```sql
CREATE OR REPLACE TABLE fintech.silver.contratos AS
SELECT
    CAST(id_contrato AS INT) AS id_contrato,
    CAST(id_cliente AS INT) AS id_cliente,
    CAST(id_produto AS INT) AS id_produto,
    CAST(data_adesao AS DATE) AS data_adesao,
    status
FROM fintech.bronze.contratos
```
